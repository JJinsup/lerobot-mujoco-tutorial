## Step 1. Configuration File: pi0_omy.yaml

Make sure `pi0_omy.yaml` is configured correctly:

```yaml
dataset:
  repo_id: omy_pnp
  root: ./omy_pnp
policy:
  type : pi0
  chunk_size: 5
  n_action_steps: 5
save_checkpoint: true
output_dir: ./ckpt/pi0_omy
batch_size: 16
job_name : pi0_omy
resume: false
seed : 42
num_workers: 8
steps: 20_000
eval_freq: -1 # No evaluation
log_freq: 50
save_checkpoint: true
save_freq: 5_000
use_policy_training_preset: true
  
wandb:
  enable: true
  project: pi0_omy
  entity: <YOUR ENTITY for wandb>
  disable_artifact: true
```

Once training is complete, use the following command to deploy and visualize the trained policy:

```bash

```

The deployment script will:
1. Load the trained model from `./ckpt/pi0_omy/checkpoints/last/pretrained_model`
2. Initialize the simulation environment
3. Run the policy in the environment with MuJoCo rendering
4. Display success/failure for each episode
